```python
# Sistema de Cadastro - Dia 5 - Menu

clientes = []

while True:
  print("\n--- MENU ---")
  print("1 - Cadastrar cliente")
  print("2 - Listar clientes")
  print("3 - Sair")

  opcao = input("Escolha uma opção: ")

  if opcao == "1":
    nome = input("Digite o nome: ")
    idade = input(f"Digite a idade de {nome}: ")
    cidade = input(f"Digite a cidade de {nome}: ")

    cliente = {"nome": nome, "idade": idade, "cidade": cidade}
    clientes.append(cliente)
    print("Cliente cadastrado!")

  elif opcao == "2":
    print("\n--- Lista de Clientes ---")
    if len(clientes)== 0:
      print("Nenhum cliente cadastradoainda.")
    else:
      for cliente in clientes:
        print(f"{cliente['nome']} - {cliente['idade']} anos - mora em {cliente['cidade']}")
      print(f"\nTotal: {len(clientes)} clientes")

  elif opcao == "3":
    print("Saindo... Até logo!")
    break

  else:
    print("Opção inválida! Tente de novo.")
```

```python
# Sistema de cadastro - Dia 6 - Buscar cliente

clientes = []

while True:
  print("\n--- MENU ---")
  print("1 - Cadastrar cliente")
  print("2 - Listar clientes")
  print("3 - Buscar cliente")
  print("4 - Sair")

  opcao = input("Escolha uma opção: ")

  if opcao == "1":
    nome = input("Digite o nome: ")
    idade = input(f"Digite a idade de {nome}: ")
    cidade = input(f"Digite a cidade de {nome}: ")

    cliente = {"nome": nome, "idade": idade, "cidade": cidade}
    clientes.append(cliente)
    print("Cliente cadastrado!")

  elif opcao == "2":
    print("\n--- Lista de Clientes ---")
    if len(clientes) == 0:
      print("Nenhum cliente cadastrado ainda.")
    else:
      for cliente in clientes:
        print(f"{cliente['nome']} - {cliente['idade']} anos - mora em {cliente['cidade']}")
      print(f"\nTotal: {len(clientes)} clientes")

  elif opcao == "3":
      busca = input("Digite o nome para buscar: ")
      encontrado = False
      for cliente in clientes:
        if cliente['nome'].lower() == busca.lower():
          print(f"nEncontrado: {cliente['nome']} - {cliente['idade']} anos - mora em {cliente['cidade']}")
          encontrado = True
      if not encontrado:
          print("Cliente não encontrado!")

  elif opcao == "4":
      print("Saindo... Até logo!")
      break

  else:
      print("Opção inválida! Tente novamente.")

```

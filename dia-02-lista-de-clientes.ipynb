```python
# Meu primeiro sistema de cadastro - Dia 1
nome = input ("Qual seu nome? ")
idade = input ("Qual sua idade? ")
cidade = input ("Onde você mora? ")

print (f"Olá {nome}, você tem {idade} anos e mora em {cidade}!")
```

```python
# Lista de clientes - Dia 2

clientes = []
for i in range (3):
    nome = input(f"Digite o nome do cliente {i+1}: ")
    clientes.append ( nome )

print ("\n--- Lista de Clientes Cadastrados ---")
for cliente in clientes:
    print (f" - {cliente} ")

print (f"\nTotal: {len(clientes)} clientes cadastrados!")
```

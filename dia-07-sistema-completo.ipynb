```python
# Sistema de Cadastro - Dia 7 - Completo

clientes = []

while True:
  print("\n--- Menu ---")
  print("1 - Cadastrar cliente")
  print("2 - Listar clientes")
  print("3 - Buscar clientes")
  print("4 - Excluir clientes")
  print("5 - Sair")

  opcao = input("Escolha uma opção: ")

  if opcao == "1":
    nome = input("Digite o nome do cliente: ")
    idade = input("Digite a idade do cliente: ")
    cidade = input("Digite a cidade do cliente: ")
    cliente = {"nome": nome, "idade": idade, "cidade": cidade}
    clientes.append(cliente)
    print ("Cliente cadastrado!")

  elif opcao == "2":
    print("\n--- Lista de Clientes ---")
    if len(clientes) == 0:
      print("Nenhum cliente cadastrado ainda.")
    else:
        for cliente in clientes:
          print(f"{cliente['nome']} - {cliente['idade']} anos, mora em {cliente['cidade']}")


  elif opcao == "3":
    busca = input("Digite o nome do cliente que deseja buscar: ")
    encontrado = False
    for cliente in clientes:
      if cliente["nome"].lower() == busca.lower():
        print (f"Cliente encontrado: {cliente['nome']} - {cliente['idade']} anos, mora em {cliente['cidade']}")
        encontrado = True
        break
    if not encontrado:
      print ("Cliente não encontrado.")

  elif opcao == "4":
    excluir = input("Digite o nome do cliente que deseja excluir: ")
    for cliente in clientes:
      if cliente["nome"].lower() == excluir.lower():
        clientes.remove(cliente)
        print("Cliente excluído com sucesso!")
        break

  elif opcao == "5":
    print ("Saindo... Até logo!")
    break
  else:
    print ("Opção inválida. Tente novamente.")





```
